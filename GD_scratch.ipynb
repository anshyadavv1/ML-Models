```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import cross_val_score
```

```python
x,y=make_regression(n_samples=100,n_features=1,n_targets=1,n_informative=1,noise=20,random_state=42)
```

```python
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(x,y)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
np.mean(cross_val_score(lr,x,y,scoring="r2",cv=10))
```

```python
class GDResgressor:

  def __init__(self,learning_rate,epochs=100):

    self.m=100
    self.b=-120
    self.lr=learning_rate
    self.epochs=epochs

  def fit(self,x,y):

    for i in range(self.epochs):
      loss_slop_b=  -2* np.sum(y-self.m*x.ravel()-self.b)
      loss_slop_m=  -2* np.sum((y-self.m*x.ravel()-self.b)*x.ravel())
      self.b=self.b-(self.lr*loss_slop_b)
      self.m=self.m-(self.lr*loss_slop_m)

    print(self.b,self.m)

  def predict(self,x):
    return self.m*x+self.b
```

```python
gd= GDResgressor(0.001,100)
```

```python
gd.fit(x,y)
```

```python
gd.predict(x)
```
